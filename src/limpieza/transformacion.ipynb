##### Análisis de componentes principales (PCA)
Lo subiremos a MinIO y analizaremos cada una de las componentes

```python
#Para subirlo a minio (ya subido)
#df_pca, pca = tf.pca()

#Para bajarlo de minio
cliente = mf.crear_cliente()
df_pca = mf.bajar_fichero(cliente, f"grupo3/cleaned/pca/final_pca_date_transformado.parquet", "df")
df_metricas = mf.bajar_fichero(cliente, f"grupo3/cleaned/pca/metricas_pca_date_transformado.parquet", "df")
```

```python
#Vemos la varianza explicada por cada componente
df_metricas.set_index(df_metricas["Componente"], inplace=True)
df_metricas.drop(["Componente"], axis = 1, inplace=True)
df_metricas
```

Vamos a ir viendo cómo va evolucionando el porcentaje de explicabilidad cada vez que vamos insertando una nueva componente para tomar la decisión de cuántos componentes incluiremos

```python
#Vemos cómo se va acumulando la varianza
varianza_total = 0
for i in range(len(df_metricas)):
    varianza_total += df_metricas["Varianza_Explicada"].iloc[i]
    print(f"{i+1} componentes explican el {round(varianza_total * 100, 2)}% de la varianza")
```
